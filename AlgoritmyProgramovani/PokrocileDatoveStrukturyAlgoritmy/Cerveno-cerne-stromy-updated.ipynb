# Červeno-černé stromy: Implementace a analýza

## 1. Úvod

Červeno-černé stromy patří mezi samovyvažující binární vyhledávací stromy. Jejich specifická vlastnost je schopnost udržovat vyváženost pomocí pravidel, která definují červenou a černou barvu jednotlivých uzlů. Tyto stromy zajišťují časovou složitost operací vkládání, mazání a vyhledávání na úrovni **O(log n)**, což je ideální pro efektivní práci s velkými datovými strukturami. 

Tato práce se zaměřuje na implementaci červeno-černého stromu a jeho analýzu, přičemž zahrnuje i srovnání s AVL stromy. Dále je diskutována časová složitost základních operací a vyvažovací mechanismy, které zajišťují efektivní výkon.

Cíle této práce:
- Implementovat červeno-černý strom s podporou operací vkládání, mazání a vyhledávání.
- Srovnat jeho vlastnosti s AVL stromy.
- Analyzovat vlastnosti stromu a časovou složitost základních operací.

---

## 2. Teoretická část

### 2.1 Červeno-černé stromy

Červeno-černé stromy jsou často používány v implementacích datových struktur, jako jsou asociativní pole (`std::map` v C++ nebo `TreeMap` v Javě), a nachází uplatnění i v databázových systémech.

#### Klíčové vlastnosti

Červeno-černé stromy splňují následující pravidla:
1. Každý uzel stromu je označen jako **červený** nebo **černý**.
2. Kořen stromu je vždy **černý**.
3. Všechny listy (NIL uzly) jsou **černé**. NIL uzly jsou reprezentovány speciálními sentinelovými uzly, které zjednodušují implementaci algoritmů.
4. Každý červený uzel má **černé potomky** (nikdy se nesetkávají dva červené uzly).
5. Všechny cesty od kořene k NIL uzlům obsahují stejný počet **černých uzlů** (tzv. černá výška stromu).

#### Vyvažovací mechanismus

Hlavním důvodem zavedení těchto pravidel je udržení maximální výšky stromu v **O(log n)**. Černá výška stromu je definována jako počet černých uzlů na cestě od kořene ke kterémukoliv listu (NIL uzlu). Tento mechanismus je klíčový pro dosažení efektivní časové složitosti všech základních operací.

#### Použití sentinelových uzlů

Sentinelové uzly představují prázdné uzly a jsou vždy černé. Díky nim je možné sjednotit implementaci algoritmů, protože není potřeba zvlášť kontrolovat, zda konkrétní uzel existuje. To zjednodušuje správu ukazatelů a ošetřování hraničních podmínek.

---

### 2.2 Porovnání červeno-černých a AVL stromů

AVL stromy, stejně jako červeno-černé stromy, patří mezi samovyvažující binární vyhledávací stromy. Hlavní rozdíly mezi těmito dvěma strukturami jsou shrnuty v následující tabulce:

| **Vlastnost**             | **Červeno-černé stromy**             | **AVL stromy**                  |
|----------------------------|---------------------------------------|----------------------------------|
| **Výška stromu**          | O(log n)                             | O(log n)                        |
| **Vyvažování**            | Méně striktní                        | Přísné                          |
| **Časová složitost operací** | Efektivní pro vkládání a mazání      | Efektivní pro vyhledávání        |
| **Použití**               | Databázové systémy, vyhledávací tabulky | Systémy s intenzivním vyhledáváním |

#### Diskuze rozdílů

Červeno-černé stromy jsou méně striktně vyvažované než AVL stromy, což jim umožňuje rychlejší provádění operací vkládání a mazání. AVL stromy naopak zaručují lepší vyváženost, což z nich činí vhodnější volbu pro aplikace, kde je klíčové minimalizovat výšku stromu. Nicméně, vyšší počet rotací u AVL stromů může způsobit vyšší režii při častých úpravách stromu.

---

### 2.3 Analýza časové složitosti

Časová složitost operací na červeno-černých stromech je omezena na **O(log n)** díky vyvažování stromu. Podrobnosti o jednotlivých operacích jsou uvedeny níže:

1. **Vkládání**:
   - Nový uzel je vždy vložen jako červený.
   - Pokud jsou porušena pravidla stromu, je vyvažování provedeno pomocí změny barev a rotací.
   - Časová složitost: **O(log n)**.

2. **Vyhledávání**:
   - Probíhá stejně jako u běžného binárního vyhledávacího stromu.
   - Časová složitost: **O(log n)**.

3. **Mazání**:
   - Mazání je složitější operace, která vyžaduje zajištění opětovného vyvážení stromu, pokud je porušena černá výška.
   - Časová složitost: **O(log n)**.

---

### Shrnutí

Červeno-černé stromy jsou všestranné datové struktury, které nachází uplatnění v širokém spektru aplikací. Díky svým vlastnostem a vyvažovacím mechanismům jsou schopné poskytovat efektivní výkon při různých operacích. Na rozdíl od AVL stromů nabízejí větší flexibilitu při úpravách dat, což je činí vhodnými pro aplikace, kde je časté vkládání a mazání dat.



## 3. Implementace
### 3.1 Červeno-černý strom


![Red-Black Tree Visualization](red_black_tree_visualization.png)

```python
%pip install termcolor
from termcolor import colored

class RedBlackNode:
    def __init__(self, key, color="red"):
        self.key = key
        self.color = color
        self.left = None
        self.right = None
        self.parent = None

class RedBlackTree:
    def __init__(self):
        self.NIL = RedBlackNode(key=None, color="black")  # Sentinel NIL node
        self.root = self.NIL

    def insert(self, key):
        new_node = RedBlackNode(key)
        new_node.left = self.NIL
        new_node.right = self.NIL
        new_node.color = "red"

        parent = None
        current = self.root

        while current != self.NIL:
            parent = current
            if new_node.key < current.key:
                current = current.left
            else:
                current = current.right

        new_node.parent = parent
        if parent is None:
            self.root = new_node
        elif new_node.key < parent.key:
            parent.left = new_node
        else:
            parent.right = new_node

        if new_node.parent is None:
            new_node.color = "black"
            return

        if new_node.parent.parent is None:
            return

        self._fix_insert(new_node)

    def _fix_insert(self, node):
        while node.parent.color == "red":
            if node.parent == node.parent.parent.right:
                uncle = node.parent.parent.left
                if uncle.color == "red":
                    uncle.color = "black"
                    node.parent.color = "black"
                    node.parent.parent.color = "red"
                    node = node.parent.parent
                else:
                    if node == node.parent.left:
                        node = node.parent
                        self._rotate_right(node)
                    node.parent.color = "black"
                    node.parent.parent.color = "red"
                    self._rotate_left(node.parent.parent)
            else:
                uncle = node.parent.parent.right
                if uncle.color == "red":
                    uncle.color = "black"
                    node.parent.color = "black"
                    node.parent.parent.color = "red"
                    node = node.parent.parent
                else:
                    if node == node.parent.right:
                        node = node.parent
                        self._rotate_left(node)
                    node.parent.color = "black"
                    node.parent.parent.color = "red"
                    self._rotate_right(node.parent.parent)
            if node == self.root:
                break
        self.root.color = "black"

    def _rotate_left(self, node):
        right_child = node.right
        node.right = right_child.left
        if right_child.left != self.NIL:
            right_child.left.parent = node

        right_child.parent = node.parent
        if node.parent is None:
            self.root = right_child
        elif node == node.parent.left:
            node.parent.left = right_child
        else:
            node.parent.right = right_child
        right_child.left = node
        node.parent = right_child

    def _rotate_right(self, node):
        left_child = node.left
        node.left = left_child.right
        if left_child.right != self.NIL:
            left_child.right.parent = node

        left_child.parent = node.parent
        if node.parent is None:
            self.root = left_child
        elif node == node.parent.right:
            node.parent.right = left_child
        else:
            node.parent.left = left_child
        left_child.right = node
        node.parent = left_child

    def search(self, key):
        return self._search_tree(self.root, key)

    def _search_tree(self, node, key):
        if node == self.NIL or key == node.key:
            return node
        if key < node.key:
            return self._search_tree(node.left, key)
        return self._search_tree(node.right, key)

    def delete(self, key):
        print(f"Funkce delete zatím není implementována pro klíč {key}")


class RedBlackTreeVisualizer(RedBlackTree):
    def print_tree(self, node, indent="", last=True):
        if node != self.NIL:
            print(indent, end="")
            if last:
                print("R----", end="")
                indent += "   "
            else:
                print("L----", end="")
                indent += "|  "

            color = "red" if node.color == "red" else "black"
            print(colored(f"{node.key} ({color})", color))
            self.print_tree(node.left, indent, False)
            self.print_tree(node.right, indent, True)


# Testovací část s vizualizací
if __name__ == "__main__":
    tree = RedBlackTreeVisualizer()

    # Vkládání uzlů a vizualizace po každém vkladu
    keys_to_insert = [20, 15, 25, 10, 5, 1, 30, 35]
    for key in keys_to_insert:
        print(f"\nVkládání klíče {key}:")
        tree.insert(key)
        tree.print_tree(tree.root)  # Vizualizace stromu po vložení

    # Vyhledávání klíčů
    keys_to_search = [10, 35, 50]
    for key in keys_to_search:
        result = tree.search(key)
        if result != tree.NIL:
            print(f"\nKlíč {key} nalezen (barva: {result.color}).")
        else:
            print(f"\nKlíč {key} nebyl nalezen.")

    # Mazání (zatím neimplementováno)
    print("\nMazání klíče 10:")
    tree.delete(10)
```

### 3.2 AVL strom


```python
class AVLNode:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None
        self.height = 1  # Výška uzlu

class AVLTree:
    def __init__(self):
        self.root = None

    def insert(self, root, key):
        # 1. Vložení uzlu do BST
        if not root:
            return AVLNode(key)
        elif key < root.key:
            root.left = self.insert(root.left, key)
        else:
            root.right = self.insert(root.right, key)

        # 2. Aktualizace výšky aktuálního uzlu
        root.height = 1 + max(self.get_height(root.left), self.get_height(root.right))

        # 3. Získání balancovacího faktoru
        balance = self.get_balance(root)

        # 4. Vyvážení stromu
        # Levá levá rotace
        if balance > 1 and key < root.left.key:
            return self.rotate_right(root)

        # Pravá pravá rotace
        if balance < -1 and key > root.right.key:
            return self.rotate_left(root)

        # Levá pravá rotace
        if balance > 1 and key > root.left.key:
            root.left = self.rotate_left(root.left)
            return self.rotate_right(root)

        # Pravá levá rotace
        if balance < -1 and key < root.right.key:
            root.right = self.rotate_right(root.right)
            return self.rotate_left(root)

        return root

    def delete(self, root, key):
        # 1. Odstranění uzlu z BST
        if not root:
            return root
        elif key < root.key:
            root.left = self.delete(root.left, key)
        elif key > root.key:
            root.right = self.delete(root.right, key)
        else:
            if not root.left:
                return root.right
            elif not root.right:
                return root.left
            temp = self.get_min_value_node(root.right)
            root.key = temp.key
            root.right = self.delete(root.right, temp.key)

        # 2. Aktualizace výšky
        root.height = 1 + max(self.get_height(root.left), self.get_height(root.right))

        # 3. Vyvážení stromu
        balance = self.get_balance(root)

        # Levá levá rotace
        if balance > 1 and self.get_balance(root.left) >= 0:
            return self.rotate_right(root)

        # Levá pravá rotace
        if balance > 1 and self.get_balance(root.left) < 0:
            root.left = self.rotate_left(root.left)
            return self.rotate_right(root)

        # Pravá pravá rotace
        if balance < -1 and self.get_balance(root.right) <= 0:
            return self.rotate_left(root)

        # Pravá levá rotace
        if balance < -1 and self.get_balance(root.right) > 0:
            root.right = self.rotate_right(root.right)
            return self.rotate_left(root)

        return root

    def rotate_left(self, z):
        y = z.right
        T2 = y.left
        y.left = z
        z.right = T2
        z.height = 1 + max(self.get_height(z.left), self.get_height(z.right))
        y.height = 1 + max(self.get_height(y.left), self.get_height(y.right))
        return y

    def rotate_right(self, z):
        y = z.left
        T3 = y.right
        y.right = z
        z.left = T3
        z.height = 1 + max(self.get_height(z.left), self.get_height(z.right))
        y.height = 1 + max(self.get_height(y.left), self.get_height(y.right))
        return y

    def get_height(self, node):
        return node.height if node else 0

    def get_balance(self, node):
        return self.get_height(node.left) - self.get_height(node.right) if node else 0

    def get_min_value_node(self, root):
        current = root
        while current.left:
            current = current.left
        return current

    def pre_order(self, root):
        if not root:
            return []
        return [root.key] + self.pre_order(root.left) + self.pre_order(root.right)

# Testování
# Vytvoření AVL stromu a vložení klíčů
avl_tree = AVLTree()
root = None

keys = [10, 20, 30, 40, 50, 25]

for key in keys:
    root = avl_tree.insert(root, key)

# Výpis AVL stromu v pre-order (pro kontrolu pořadí uzlů)
print("Pre-order průchod AVL stromu:")
print(avl_tree.pre_order(root))

# Odstranění klíče a nový průchod stromem
root = avl_tree.delete(root, 20)
print("Pre-order průchod AVL stromu po odstranění klíče 20:")
print(avl_tree.pre_order(root))
```

## Porovnání implementací a výstupů červeno-černých a AVL stromů

### Implementace

1. **Červeno-černé stromy:**
   - **Vyvažování:** Méně striktní, umožňuje dočasnou nerovnováhu, což vede k menšímu počtu rotací při vkládání a mazání.
   - **Barvy uzlů:** Každý uzel je buď červený, nebo černý. Barvy se využívají k zajištění rovnováhy.
   - **Sentinelové uzly:** NIL uzly slouží k zjednodušení implementace a vždy mají černou barvu.

   #### Výhody:
   - Menší režie při častých operacích (vkládání, mazání).
   - Stabilní časová složitost **O(log n)**.

   #### Nevýhody:
   - Složitější implementace kvůli správě barev a speciálním pravidlům.

2. **AVL stromy:**
   - **Vyvažování:** Přísné, zajišťuje, že rozdíl výšek podstromů je vždy maximálně 1.
   - **Rotace:** Vyžaduje více rotací při vkládání a mazání než červeno-černé stromy.
   - **Výška stromu:** Menší než u červeno-černých stromů, což zajišťuje rychlejší vyhledávání.

   #### Výhody:
   - Optimální pro operace s intenzivním vyhledáváním.
   - Jednodušší implementace než červeno-černé stromy.

   #### Nevýhody:
   - Vyšší režie při operacích vkládání a mazání.

---

### Výstupy

#### Červeno-černé stromy:
- **Vkládání:** Udržují rovnováhu pomocí barev a rotací. Každý krok vede ke konzistentnímu stromu splňujícímu pravidla červeno-černých stromů.
- **Mazání:** Není implementováno v aktuální verzi kódu, ale probíhá obdobně jako vkládání s dodatečnou kontrolou rovnováhy.
- **Vyhledávání:** Stabilní časová složitost **O(log n)**, díky zajištění maximální výšky stromu.

#### AVL stromy:
- **Vkládání:** Každá operace vyžaduje kontrolu rozdílu výšek podstromů. Pokud je rozdíl větší než 1, provádí se rotace.
- **Mazání:** Po odstranění uzlu se strom přeuspořádá, aby splňoval podmínku maximálního rozdílu výšek podstromů.
- **Vyhledávání:** Rychlejší než u červeno-černých stromů díky menší výšce stromu.

---

### Shrnutí

1. **Výhody červeno-černých stromů:**
   - Rychlejší při operacích vkládání a mazání.
   - Vhodné pro aplikace, kde dochází k častým úpravám dat (např. databázové systémy).

2. **Výhody AVL stromů:**
   - Lepší pro intenzivní vyhledávání díky menší výšce stromu.
   - Stabilní a jednodušší vyvažování založené na výšce podstromů.

3. **Volba stromu:**
   - **Červeno-černé stromy**: Pokud aplikace vyžaduje časté úpravy dat.
   - **AVL stromy**: Pokud je prioritou rychlé vyhledávání.

Tato analýza ukazuje, že volba mezi červeno-černými a AVL stromy závisí na konkrétních požadavcích aplikace. Obě datové struktury jsou však efektivní a vhodné pro široké spektrum použití.
